```python
import matplotlib.pyplot as plt

print("Analytics 18: Height vs Weight Scatter Plot")
plt.scatter(df['height'], df['weight'])
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
plt.hist(df['weight'])
plt.show()
```

```python
avg_height = df.groupby("cardio")["height"].mean()

plt.bar(avg_height.index, avg_height.values)
plt.xlabel("Cardiovascular Disease")
plt.ylabel("Average Height (cm)")
plt.title("Average Height by Cardiovascular Disease")
plt.xticks([0, 1], ["No Disease", "Disease"])

plt.show()
```

```python
import matplotlib.pyplot as plt

cardio_counts = df["cardio"].value_counts()

plt.pie(
    cardio_counts,
    labels=["No Disease", "Disease"],
    autopct="%1.1f%%",
    # startangle=90
)

plt.title("Distribution of Cardiovascular Disease")
plt.show()
```
